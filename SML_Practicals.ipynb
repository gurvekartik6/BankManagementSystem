```python
# Practical 6
def discriminant(x1, x2, w1, w2, b):
    return w1 * x1 + w2 * x2 + b

x1 = float(input("Enter first feature: "))
x2 = float(input("Enter second feature: "))

score1 = discriminant(x1, x2, 2, 1, -3)
score2 = discriminant(x1, x2, -1, 2, 1)

if score1 > score2:
    print("Class 1")
else:
    print("Class 2")
```

```python
# Practical 7
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

data = load_breast_cancer()
x = data.data
y = data.target

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

model = LogisticRegression(max_iter=5000)
model.fit(x_train, y_train)

prob = model.predict_proba(x_test)[:, 1]

fpr, tpr, thresholds = roc_curve(y_test, prob)
score = auc(fpr, tpr)

plt.plot(fpr, tpr, label="AUC = " + str(round(score, 2)))
plt.plot([0, 1], [0, 1], "--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
# Practical 8

import numpy as np
import matplotlib.pyplot as plt

data = np.array([2, 3, 3, 4, 5, 6, 7, 8])
width = 1.0

def estimate_density(x):
    total = 0

    for value in data:
        z = (x - value) / width
        total += np.exp(-0.5 * z * z)

    return total / (len(data) * width * np.sqrt(2 * np.pi))

points = np.linspace(0, 10, 100)
density = [estimate_density(x) for x in points]

plt.plot(points, density)
plt.xlabel("Data Values")
plt.ylabel("Probability Density")
plt.title("Parzen Window Density Estimation")
plt.show()
```

```python
# Practical 10
import numpy as np

x = np.array([[1, 2], [2, 3], [3, 2],
              [7, 8], [8, 7], [9, 8]])

y = np.array([0, 0, 0, 1, 1, 1])

def predict(point):
    classes = np.unique(y)
    scores = []

    for c in classes:
        samples = x[y == c]
        distance = np.sum((samples - point) ** 2, axis=1)
        score = np.sum(np.exp(-distance / 2))
        scores.append(score)

    return classes[np.argmax(scores)]

point = np.array([6, 4])
print("Predicted Class:", predict(point))
```
